**Exercício 2.2**

```python
#Exercício 2.2
d = int(input('Quantos dias?: '))
h = int(input('Quantas horas?: '))
m = int(input('Quantos minutos?: '))
s = int(input('Quantos segundos?: '))
#1 dia = 24h | 1h = 60 min | 1 min = 60 s
total = s + (m*60) + (h*60*60)+(d*24*60*60)
#Maneira clássica
res = 'O total de segundos calculado é de %i.'%total
print(res)
#Maneira moderna
res = 'O total de segundos calculado é de {}.'.format(total)
print(res)
#Maneira com f-string
res = f'O total de segundos calculado é de {total}.'
print(res)
```

```python
um_minuto = 60
uma_hora = 3600
um_dia = 86400
dias = int(input('Digite quantos dias: '))
horas = int(input('Digite quantas horas: '))
minutos = int(input('Digite quantos minutos: '))
segundos = int(input('Digite quantos segundos: '))
total = dias * um_dia + horas * uma_hora + minutos * um_minuto + segundos
print(f'O total de segundos é {total}')
```

**Exercício 2.3**

```python
#Exercício 2.3
preco = float(input('Digite o preço do produto: '))
p = float(input('Digite o percentual de desconto(o-100%): '))
desconto = preco * p/100
final = preco - desconto
#Maneira clássica
print('O preço do produto é %.2f. Desconto de %.0f%%.'% (preco, p))
print('Valor calculado de desconto: %.2f. Valor final do produto: %.2f' % (desconto, final))
#Maneira moderna
print('O preço do produto é {}. Desconto de {}%.'.format(preco, p))
print('Valor calculado de desconto: {}. Valor final do produto: {}'.format(desconto, final))
```

```python
preco = float(input('Digite o preço do produto: '))
percentual_desconto = float(input('Digite o percentual de desconto: '))
desconto = preco * percentual_desconto/100
preco_final = preco - desconto
print(f'O desconto é de {desconto}')
print(f'O preço final é {preco_final}')
```
