**TEMA 2 – PARÂMETROS EM FUNÇÕES**

```python
def realce(s1):
  print('|','  ' * 10,'|')
  print('|','__' * 10,'|')
  print(s1)
  print('|','__' * 10,'|')
  print('|','__' * 10,'|')
```

```python
def soma3(x, y, z):
  res = x + y + z
  print(res)
```

```python
def soma3(x = 0, y = 0, z = 0):
  res = x + y + z
  print(res)
```

```python
soma3(1,2,3)
soma3(1,2) #z foi omitido
soma3(1)   #y e z foram omitidos
soma3()    #x, y e z foram omitidos
```

```python
def soma3(x = 0, y=0, z=0, imprime=False):
  res = x + y + z
  if imprime:
    print(res)
```
